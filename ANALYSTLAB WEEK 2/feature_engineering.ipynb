```python
# FEATURE ENGINEERING
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt


# LOAD THE ORIGINAL DATASET
df = pd.read_csv("/content/WA_Fn-UseC_-Telco-Customer-Churn.csv")

print("Dataset loaded successfully.")

print("\nOriginal dataset shape:")
print(df.shape)


# DISPLAY ORIGINAL DATA

print("\nFIRST 10 ROWS OF ORIGINAL DATASET")

display(df.head(10))

# Make a copy to work on for cleaning and feature engineering
clean_df = df.copy()

# DATA CLEANING AND TYPE CONVERSION

# Convert 'TotalCharges' to numeric, handling missing values
# 'TotalCharges' contains some empty strings, which should be NaN
clean_df['TotalCharges'] = pd.to_numeric(clean_df['TotalCharges'], errors='coerce')
# Fill NaN values (e.g., from empty strings or new customers with 0 tenure) with 0
clean_df['TotalCharges'] = clean_df['TotalCharges'].fillna(0)

print("\n'TotalCharges' column processed and converted to numeric.")


# ENCODING CATEGORICAL FEATURES
# Convert target variable 'Churn': 'Yes' -> 1, 'No' -> 0
clean_df['Churn'] = clean_df['Churn'].map({'Yes': 1, 'No': 0})
print("\n'Churn' column encoded.")

# Convert 'gender': 'Female' -> 0, 'Male' -> 1
clean_df['gender'] = clean_df['gender'].map({'Female': 0, 'Male': 1})
print("\n'gender' column encoded.")

# Binarize other 'Yes'/'No' columns
binary_columns = [
    'Partner', 'Dependents', 'PhoneService', 'PaperlessBilling'
]
for col in binary_columns:
    clean_df[col] = clean_df[col].map({'Yes': 1, 'No': 0})
print("Binary 'Yes'/'No' columns encoded.")

# Handle service-related columns with 'No internet service' or 'No phone service'
service_columns = [
    'MultipleLines', 'OnlineSecurity', 'OnlineBackup', 'DeviceProtection',
    'TechSupport', 'StreamingTV', 'StreamingMovies'
]

for col in service_columns:
    # Replace 'No phone service' or 'No internet service' with 'No'
    # Then map 'Yes' -> 1, 'No' -> 0
    clean_df[col] = clean_df[col].replace({'No phone service': 'No', 'No internet service': 'No'})
    clean_df[col] = clean_df[col].map({'Yes': 1, 'No': 0})
print("Service-related columns encoded.")

# One-Hot Encode multi-category nominal features
# 'customerID' is an identifier and will be dropped later
categorical_columns_to_encode = [
    'InternetService', 'Contract', 'PaymentMethod'
]

clean_df = pd.get_dummies(clean_df, columns=categorical_columns_to_encode, drop_first=False)
print("Multi-category nominal features one-hot encoded.")


#REMOVE IRRELEVANT IDENTIFIERS


# Drop 'customerID' as it is a unique identifier and not useful for modeling
clean_df = clean_df.drop(columns=['customerID'])
print("\n'customerID' column removed.")


# FINAL DATASET STRUCTURE CHECK

print("\nFINAL CLEANED DATASET INFO")
clean_df.info()

print("\nFINAL DATASET SHAPE")
print(clean_df.shape)

print("\nFINAL DATASET HEAD")
display(clean_df.head())


# SAVE FEATURE-ENGINEERED DATASET


output_file = "Telco_Customer_Churn_Feature_Engineered.csv"

clean_df.to_csv(output_file, index=False)

print("\nFeature-engineered dataset saved as:", output_file)

print("\nOriginal dataset columns (excluding customerID, if dropped by original code):", len(df.columns))
print("Final processed dataset columns:", len(clean_df.columns))

print("\nThe feature-engineered Telco Customer Churn dataset is ready for further steps like scaling and model training.")
```
