```python
# IMPORT LIBRARIES
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split

# LOAD THE FINAL DATASET FROM PART 5
df = pd.read_csv(
    "/content/WA_Fn-UseC_-Telco-Customer-Churn.csv"
)
print("Final dataset loaded successfully.")

# DISPLAY FIRST 10 ROWS
print("\nFIRST 10 ROWS")
display(
    df.head(10)
)

# CHECK DATASET DIMENSIONS
print("\nDATASET DIMENSIONS")
print(
    "Rows:",
    df.shape[0]
)
print(
    "Columns:",
    df.shape[1]
)

# CHECK DATA TYPES
print("\nDATA TYPES")
print(
    df.dtypes
)

# CHECK MISSING VALUES
print("\nMISSING VALUES")
missing_values = df.isnull().sum()
print(
    missing_values
)
print(
    "\nTotal missing values:",
    missing_values.sum()
)

# CHECK DUPLICATES
print("\nDUPLICATE RECORDS")
duplicate_count = df.duplicated().sum()
print(
    "Number of duplicate rows:",
    duplicate_count
)

# CHECK TARGET VARIABLE
print("\nTARGET VARIABLE: CHURN")
print(
    df["Churn"].value_counts()
)

print("\nTARGET VARIABLE PERCENTAGES")
print(
    df["Churn"]
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)

# CHECK FOR INFINITE VALUES
print("\nINFINITE VALUES")
infinite_values = np.isinf(
    df.select_dtypes(
        include=np.number
    )
).sum().sum()
print(
    "Number of infinite values:",
    infinite_values
)

# CHECK DESCRIPTIVE STATISTICS
print("\nDESCRIPTIVE STATISTICS")
display(
    df.describe()
)

# SEPARATE FEATURES AND TARGET
X = df.drop(
    "Churn",
    axis=1
)
y = df["Churn"]

print("\nFEATURE MATRIX")
print(
    "X shape:",
    X.shape
)

print("\nTARGET")
print(
    "y shape:",
    y.shape
)

# VERIFY ALL FEATURES ARE NUMERICAL
non_numeric_columns = X.select_dtypes(
    exclude=np.number
).columns.tolist()

print("\nNON-NUMERICAL FEATURES")
print(
    non_numeric_columns
)

if len(non_numeric_columns) == 0:
    print(
        "All predictor variables are numerical."
    )
else:
    print(
        "Warning: Some predictor variables are still non-numerical."
    )

# VERIFY TARGET VARIABLE
print("\nTARGET DATA TYPE")
print(
    y.dtype
)

print("\nTARGET UNIQUE VALUES")
print(
    y.unique()
)

# TRAIN-TEST SPLIT
# Split the data into training and testing sets.
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

# DISPLAY TRAINING AND TESTING SHAPES
print("\nTRAINING DATA")
print(
    "X_train:",
    X_train.shape
)
print(
    "y_train:",
    y_train.shape
)

print("\nTESTING DATA")
print(
    "X_test:",
    X_test.shape
)
print(
    "y_test:",
    y_test.shape
)

# CHECK TARGET DISTRIBUTION
print("\nTRAINING TARGET DISTRIBUTION")
print(
    y_train.value_counts(
        normalize=True
    ).mul(100).round(2)
)

print("\nTESTING TARGET DISTRIBUTION")
print(
    y_test.value_counts(
        normalize=True
    ).mul(100).round(2)
)

# SAVE TRAINING DATA
train_df = X_train.copy()
train_df["Churn"] = y_train.values
train_df.to_csv(
    "HR_Employee_Churn_Train.csv",
    index=False
)

# SAVE TESTING DATA
test_df = X_test.copy()
test_df["Churn"] = y_test.values
test_df.to_csv(
    "HR_Employee_Churn_Test.csv",
    index=False
)

# SAVE FINAL MACHINE-LEARNING DATASET AGAIN
df.to_csv(
    "HR_Employee_Churn_Final_ML_Ready.csv",
    index=False
)

# FINAL VALIDATION
print("\nFINAL DATASET VALIDATION")
print(
    "\nNumber of rows:",
    df.shape[0]
)
print(
    "Number of columns:",
    df.shape[1]
)
print(
    "Missing values:",
    df.isnull().sum().sum()
)
print(
    "Duplicate rows:",
    df.duplicated().sum()
)
print(
    "Infinite values:",
    infinite_values
)
print(
    "Non-numerical predictor columns:",
    len(non_numeric_columns)
)

# FINAL MACHINE-LEARNING READINESS CHECK
if (
    df.isnull().sum().sum() == 0
    and
    df.duplicated().sum() == 0
    and
    infinite_values == 0
    and
    len(non_numeric_columns) == 0
):
    print(
        "\nSTATUS: DATASET IS MACHINE-LEARNING READY."
    )
else:
    print(
        "\nSTATUS: DATASET REQUIRES FURTHER PREPROCESSING."
    )

# FINAL SUMMARY
print("\nWEEK 2 - PART 6 COMPLETED")
print(
    "\nFinal ML-ready dataset:"
)
print(
    "HR_Employee_Churn_Final_ML_Ready.csv"
)
print(
    "\nTraining dataset:"
)
print(
    "HR_Employee_Churn_Train.csv"
)
print(
    "\nTesting dataset:"
)
print(
    "HR_Employee_Churn_Test.csv"
)
print(
    "\nThe dataset is now ready for machine-learning model development."
)
```
