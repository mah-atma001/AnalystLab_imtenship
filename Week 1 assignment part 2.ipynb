```python
import pandas as pd

df = pd.read_csv("/content/drive/MyDrive/WA_Fn-UseC_-Telco-Customer-Churn.csv")

initial_rows = len(df)
df = df.dropna(how="all")
removed_rows = initial_rows - len(df)
if removed_rows:
    print(f"Removed {removed_rows} fully empty rows from the dataset.")

print("\nFIRST 5 ROWS OF THE DATASET")
print("---------------------------")
print(df.head())

print("\nDATASET DIMENSIONS")
print("------------------")
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

print("\nCOLUMN NAMES")
print("------------")
for column in df.columns:
    print(column)

print("\nDATA TYPES")
print("----------")
print(df.dtypes)

print("\nDATASET INFORMATION")
print("-------------------")
df.info()

print("\nMISSING VALUES BY COLUMN")
print("------------------------")
missing_values = df.isnull().sum()
print(missing_values)

print("\nTotal missing values:", missing_values.sum())

print("\nDUPLICATE ROWS")
print("--------------")
duplicates = df.duplicated().sum()
print("Number of duplicate rows:", duplicates)

print("\nDESCRIPTIVE STATISTICS - NUMERICAL VARIABLES")
print("---------------------------------------------")
print(df.describe())

print("\nDESCRIPTIVE STATISTICS - CATEGORICAL VARIABLES")
print("-----------------------------------------------")
category_cols = df.select_dtypes(include="object").columns
if len(category_cols) > 0:
    print(df.describe(include="object"))
else:
    print("No categorical columns found in the dataset to describe.")

print("\nATTRITION/CHURN COUNTS")
print("------------------------")
if "Attrition" in df.columns:
    target_col = "Attrition"
elif "Churn" in df.columns:
    target_col = "Churn"
else:
    target_col = None

if target_col is not None:
    counts = df[target_col].value_counts()
    percentages = df[target_col].value_counts(normalize=True) * 100
    print(f"Using target column: {target_col}")
    print(counts)
    print("\nTarget percentages")
    print(percentages)
else:
    print("The dataset does not contain 'Attrition' or 'Churn' columns.")

print("\nDATA INSPECTION SUMMARY")
print("-----------------------")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
print("Total missing values:", missing_values.sum())
print("Duplicate rows:", duplicates)
if target_col is not None:
    print(f"Target column used: {target_col}")
print("\nData inspection completed successfully.")
```
